Series

```python
import pandas as pd
data=pd.Series([11,10,340,23,13])               # by default index is 0,1,2,....
data

```

```python
data=[11,10,340,23,13]                               #changes the index of the series
l=pd.Series(data,index=["a","b","c","d","e"])
l
```

```python
l.loc["a"]      #returns the value at that location 
```

```python
l.iloc[3]             #returns the value at that index location
```

```python
l[l>200]             #prints the value in l that is greater than 200
```

DataFrames

```python
data = {                                        
    "Name": ["Ram", "Hari", "Sita"],
    "Age": [20, 21, 22],
    "City": ["Pokhara", "Kathmandu", "Butwal"]
}                                                       #Treats each dictionary as a column

df = pd.DataFrame(data)
df
```

```python
df.loc[2]             #prints all the information related to the index
```

```python
df["Job"]=["Cook","Cashier","Teacher"]  #add new column
df
```

```python
new_row=pd.DataFrame([{"Name":"Prasiddha","Age":18,"City":"Pokhara","Job":"Student"}],index=[3])
df=pd.concat([df, new_row])       #add new row
df
```
