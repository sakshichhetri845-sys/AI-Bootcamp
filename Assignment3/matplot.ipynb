 ScatterPlot

```python
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib as mpl
import seaborn as sns
iris=pd.read_csv('iris.csv')
stocks=pd.read_csv('stocks-monthly.csv',parse_dates=[0])

```

```python
plt.scatter(iris['sepal_width'],iris['sepal_length'])
plt.xlabel('Sepal Width')    #name the X-axis
plt.ylabel('Sepal length')      #name the Y axis
plt.title('Scatterplot')        #name the plot
plt.grid()                      #add gridlines
```

```python
mpl.colors.cnames             #all supported colors in matplotlib
```

```python
plt.scatter(iris['sepal_width'],iris['sepal_length'],c= '#000000') #change the colour
```

```python
mpl.markers.MarkerStyle.markers  #all available markers
```

```python
#Scatterplot using seaborn
sns.scatterplot(data=iris,x="sepal_width",y="sepal_length", hue ="Species").set(title="Scatterplot")
```

BAR PLOT

```python
plt.bar(stocks['Date'].dt.month, stocks['GOOG'])
```

```python
plt.bar(stocks['Date'].dt.month, stocks['GOOG']) #create the bar plot using the months and closing prices of Google
plt.legend(['GOOG'])    #add legend
plt.xlabel('Month')            #name the X axis
plt.ylabel('Closing price')    #name the y axis
plt.title("Bar plot")          #name the plot
plt.grid()                     #add the gridlines
```

```python
#Using Seaborn
stocks["Month"]=stocks["Date"].dt.month
sns.barplot(data=stocks, x="Month",y="GOOG",hue='Month').set(title="Bar plot")
```

Stacked Bar Chart    #no direct function for plotting stacked bar chart

```python
plt.bar(stocks['Date'].dt.month, stocks['GOOG'], color='r')
plt.bar(stocks['Date'].dt.month, stocks['MSFT'], bottom=stocks['GOOG'], color='b')
plt.bar(stocks['Date'].dt.month, stocks['AMZN'], bottom=stocks['GOOG']+stocks['MSFT'], color='y')
plt.bar(stocks['Date'].dt.month, stocks['NASDAQ'], bottom=stocks['GOOG']+stocks['MSFT']+stocks['AMZN'], color='g')
plt.xlabel("Months")
plt.ylabel("Closing price")
plt.legend(["GOOG", "MSFT", "AMZN", "NASDAQ"])
plt.title("Closing prices of stocks in each month")
plt.ylim(0,800)        #add limit for Y axis to better  visualize all the catagories
```

Line Chart

```python
#Using Seaborn
sns.lineplot(data = stocks, x = 'Date', y = 'NASDAQ').set(title="Line plot")
```

```python
#Using Matplotlib
stocks_d = stocks.set_index('Date')           #first set the date column as index
stocks_d.plot()                               #create line plot with title
plt.title("Stock prices over time")
```

```python
fig, ax = plt.subplots(nrows=2, ncols=2, squeeze=False, sharex=True, figsize=(10,10))   #dimension id 2x2 and same x axis is shared

# NASDAQ
stocks['NASDAQ'].plot(ax=ax[0, 0])             #top left corner
ax[0, 0].set_title('NASDAQ')
ax[0, 0].set_ylabel('Price (USD)')

# AMZN
stocks['AMZN'].plot(ax=ax[0, 1])               #top right corner
ax[0, 1].set_title('AMZN')

# GOOG
stocks['GOOG'].plot(ax=ax[1, 0])               #down left corner
ax[1, 0].set_title('GOOG')
ax[1, 0].set_ylabel('Price (USD)')

# MSFT
stocks['MSFT'].plot(ax=ax[1, 1])                #down right corner
ax[1, 1].set_title('MSFT')

# Whole figure title
fig.suptitle("Development of stocks over time", size=18, weight='bold')
```

Heat Map

```python
#first rearrange the data to create a pivot table
iris=pd.read_csv('iris.csv')
levels = ["tiny", "small", "medium", "big", "large"]
iris["petal_width_level"] = pd.cut(iris["petal_width"], len(levels), labels=levels)
iris["petal_length_level"] = pd.cut(iris["petal_length"], len(levels), labels=levels)

iris_grouped = iris.groupby(["petal_width_level", "petal_length_level"]).count().reset_index()

# fill the NaN values with 0's
iris_grouped["count"] = iris_grouped["Species"].fillna(0)

# pivot the table
iris_matrix = iris_grouped.pivot(index="petal_width_level",columns= "petal_length_level",values= "count")

# pivot orders the levels alphabetically, so reorder them according to the order in the 'levels' variable
iris_matrix = iris_matrix.reindex(levels, axis=0);
iris_matrix = iris_matrix.reindex(levels, axis=1);

iris_matrix

```

```python

plt.imshow(iris_matrix)
plt.colorbar()
plt.title("Heatmap with color bar")
```

```python
sns.heatmap(iris_matrix, square=True).set(title="Heatmap with color bar")
```
